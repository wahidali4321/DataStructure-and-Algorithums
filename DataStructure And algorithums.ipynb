## Create an algorithm to find the lowest value in a list

```python
my_array = [7, 12, 9, 4, 11, 8]
minVal = my_array[0]

for i in my_array:
  if i < minVal:
    minVal = i

print('Lowest value:', minVal)
```

## repeated

```python
my_array = [1,2,3,4,5,6]
minVal = my_array[0]

for i in my_array:
    if i < minVal:
        minVal = i

print("The lowest value is : " , minVal)
```

```python
my_array = [1,2,34,45,56,67]
minVal = my_array[0]

for i in my_array:
    if i < minVal:
        minVal = i

print(minVal)
```
